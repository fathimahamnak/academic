```python
currentyear=int(input("Enter current year:"))
finalyear=int(input("Enter final year:"))
for year in range(currentyear,finalyear+1):
    if year % 400 == 0 or (year % 4 == 0 and year % 100!=0):
        print("Future leap year is:", year)
```

```python
import datetime
currentyear=datetime.date.today().year
finalyear=int(input("Enter final year:"))
for year in range(currentyear,finalyear+1):
    if year % 400 == 0 or (year % 4 == 0 and year % 100!=0):
        print("Future leap year is:",year)
```

```python
numbers=[-3,-1,0,2,4,]
positive=[i for i in numbers if i>0]
print("positive list:",positive)

square=[i * i for i in numbers]
print("square of list:",square)
```

```python
word=input("Enter a word:")
vowels=[i for i in word if i in "aeiou"]
print("Vowels",vowels)

values=[ord(i) for i in word]
print("Ordinal values:",values)
```

```python
num=int(input("Enter a number:"))
if num%7==0:
    print(num,"is visible by 7")
else:
    print(num,"is not visible by 7")
```

```python
ch=str(input("Enter a character:"))
vowels="a,e,i,o,u,A,E,I,O,U"
if ch in vowels:
    print(ch,"is a vowel")
else:
    print(ch,"is not a vowel")
```

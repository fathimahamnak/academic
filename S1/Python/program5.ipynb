```python
l1=list(map(int,input("Enter first list:").split()))
l2=list(map(int,input("Enter second list:").split()))
if len(l1)==len(l2):
    print("Both lists are of same length")
else:
    print("Both lists are not of same length")
    
if sum(l1)==sum(l2):
    print("Both lists have same sum")
else:
    print("Both lists not have same sum")
    
if set(l1)&set(l2):
    print("Both occur same value:",set(l1)&set(l2))
else:
    print("Both not occur same value")
```
